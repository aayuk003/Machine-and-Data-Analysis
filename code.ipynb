```python
"""
Program 2: Demonstrate the Data Analytics Lifecycle using CRISP-DM methodology
"""
 
import pandas as pd
 
 
class CRISP_DM:
    def __init__(self, data):
        self.data = data
        self.model_result = None
 
    def business_understanding(self):
        print("\n1. BUSINESS UNDERSTANDING")
        print("Goal: Predict whether a machine will fail based on temperature and vibration.")
 
    def data_understanding(self):
        print("\n2. DATA UNDERSTANDING")
        print(self.data.head())
        print("\nSummary statistics:")
        print(self.data.describe())
 
    def data_preparation(self):
        print("\n3. DATA PREPARATION")
        self.data = self.data.dropna()
        self.data["risk_score"] = (self.data["temperature"] * 0.6 +
                                    self.data["vibration"] * 0.4)
        print("Added 'risk_score' feature. Cleaned data shape:", self.data.shape)
 
    def modeling(self):
        print("\n4. MODELING")
        threshold = self.data["risk_score"].mean()
        self.data["predicted_failure"] = self.data["risk_score"] > threshold
        self.model_result = self.data
        print(f"Simple threshold model built. Threshold = {threshold:.2f}")
 
    def evaluation(self):
        print("\n5. EVALUATION")
        failure_rate = self.model_result["predicted_failure"].mean() * 100
        print(f"Predicted failure rate: {failure_rate:.2f}%")
 
    def deployment(self):
        print("\n6. DEPLOYMENT")
        print("Model logic can be deployed as a rule in the monitoring dashboard:")
        print("IF risk_score > threshold THEN flag machine for maintenance")
 
 
if __name__ == "__main__":
    sample_data = pd.DataFrame({
        "machine_id": range(1, 11),
        "temperature": [70, 85, 90, 60, 95, 55, 80, 100, 65, 75],
        "vibration": [3, 5, 6, 2, 7, 1, 4, 8, 2, 3]
    })
 
    pipeline = CRISP_DM(sample_data)
    pipeline.business_understanding()
    pipeline.data_understanding()
    pipeline.data_preparation()
    pipeline.modeling()
    pipeline.evaluation()
    pipeline.deployment()
```
